```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df = pd.read_csv("news.csv")
df = df[['text', 'label']]
df.head()
```

```python
x = df['text']
y = df['label']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
vectorizer = TfidfVectorizer(stop_words='english', max_df=0.7)
tfidf_train = vectorizer.fit_transform(x_train)
tfidf_test = vectorizer.transform(x_test)
```

```python
model = PassiveAggressiveClassifier(max_iter=50)
model.fit(tfidf_train, y_train)
```

```python
y_pred = model.predict(tfidf_test)
score = accuracy_score(y_test, y_pred)
print(f"Accuracy: {round(score * 100, 2)}%")

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)
```

```python
def predict_news(text):
    vec = vectorizer.transform([text])
    prediction = model.predict(vec)
    return prediction[0]

sample_text = "The government confirms the new vaccine is 100% effective."
print("Prediction:", predict_news(sample_text))
```

```python
# Function to predict if the news is REAL or FAKE
def predict_news(text):
    vec = vectorizer.transform([text])  # Vectorize the input
    prediction = model.predict(vec)     # Make prediction
    return prediction[0]

# Sample news articles to test
sample_news = [
    "The government has launched a new healthcare initiative for rural areas.",
    "Aliens landed in Chennai and danced on Marina Beach.",
    "NASA discovers oxygen in Mars atmosphere.",
    "Drinking cola daily boosts your intelligence."
]

# Display predictions
for i, news in enumerate(sample_news, 1):
    print(f"{i}. {news}")
    print("→ Prediction:", predict_news(news))
    print()
```
